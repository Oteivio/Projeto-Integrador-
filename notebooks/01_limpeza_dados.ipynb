# 📓 Jupyter Notebook 01: Limpeza e Higienização de Dados
### Projeto Integrador — Análise de Vendas Globais da BMW (2018-2025)

Este notebook executa o pipeline inicial de **Carga, Entendimento e Higienização** dos dados brutas de vendas da BMW. O objetivo é ler a base de dados integrada na pasta `data/`, realizar diagnósticos estatísticos descritivos e certificar que a base está pronta para ser importada na ferramenta de visualização (Power BI/Tableau).

```python
import pandas as pd
import numpy as np
import os

print("✅ Bibliotecas de ciência de dados (Pandas & NumPy) carregadas com sucesso!")
```

### 📥 1. Carregamento dos Dados
Buscamos o dataset local `./data/bmw_global_sales_2018_2025.csv` utilizando caminhos relativos de diretórios.

```python
# Definir o caminho relativo para o arquivo de dados original
caminho_csv = os.path.join("..", "data", "bmw_global_sales_2018_2025.csv")

# Carregar a base de dados usando o Pandas
df = pd.read_csv(caminho_csv)

# Mostrar o tamanho da base carregada
print(f"📊 Base carregada com sucesso!")
print(f"   - Quantidade de Linhas (Registros): {df.shape[0]}")
print(f"   - Quantidade de Colunas (Atributos): {df.shape[1]}")
```

### 🔍 2. Inspeção Inicial e Análise Descritiva
Vamos examinar as primeiras 5 linhas e verificar o tipo de dado de cada coluna.

```python
# Visualizar as primeiras 5 linhas do dataset para inspeção visual inicial
df.head()
```

```python
# Informações do DataFrame (nomes das colunas, contagem de valores não nulos e tipos de dados)
df.info()
```

### 🧹 3. Verificação de Valores Nulos e Duplicados
Essencial para garantir o rigor acadêmico da análise de dados e mitigar erros em cálculos agregados de faturamento ou médias de vendas.

```python
# Contagem de valores nulos/omissos em cada coluna
print("❌ Contagem de valores nulos por coluna:")
print(df.isnull().sum())

# Contagem de linhas duplicadas redundantes na base
duplicados = df.duplicated().sum()
print(f"\n📝 Linhas redundantes/duplicadas na base: {duplicados}")
```

### ⚙️ 4. Estatística Descritiva das Variáveis Numéricas
Compreensão das amplitudes, desvios padrões, médias e quartis das métricas.

```python
# Estatística descritiva das colunas numéricas (Year, Month, Units_Sold, Avg_Price_EUR, Revenue_EUR, BEV_Share, Premium_Share, GDP_Growth, Fuel_Price_Index)
df.describe()
```

### 💾 5. Exportação da Base Higienizada
Salvamos a base final limpa para a pasta `/data` para alimentar diretamente o Power BI ou o Tableau.

```python
caminho_saida = os.path.join("..", "data", "bmw_global_sales_2018_2025_cleaned.csv")

# Exportar para CSV sem exportar o índice automático do Pandas
df.to_csv(caminho_saida, index=False)
print(f"🎉 Base tratada e higienizada exportada com sucesso em:")
print(f"   📁 {caminho_saida}")
```
